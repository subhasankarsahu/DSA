```python
# Factorial of a number

def factorial(n):
    if n == 0:
        return 1
    
    return n * factorial(n - 1)

print(factorial(5))
```

```python
# Sum of N Numbers

def sum_n(n):
    if n == 0:
        return 0
    
    return n + sum_n(n - 1)

print(sum_n(5))
```

```python
# Fibonacci 

def fibonacci(n):
    if n <= 1:
        return n
    
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(3))
```

```python
# Reverse a String

def reverse_string(s):
    if len(s) == 0:
        return s
    
    return reverse_string(s[1:]) + s[0]

print(reverse_string("Hello I am Subha"))
```

```python
# Sum of first n numbers

def func(n):
    if n==0:
        return 0
    return n + func(n - 1)

print(func(5))
```

```python
# Reverse a number

def reverse(arr, i = 0):
    if i >= len(arr) // 2:
        return
    
    arr[i], arr[-i-1] = arr[-i-1], arr[i]
    reverse(arr, i + 1)

arr = [1,2,3,4,5]
reverse(arr)
print(arr)
```

```python
# Check for isPalindrome

def is_Palindrome(s, i=0):
    if (i>=len(s) // 2):
        return True
    if (s[i] != s[len(s)-i-1] ):
        return False
    return is_Palindrome(s, i+1)

s = "MADAM"
print(is_Palindrome(s))
```

```python
# Fibonacci using multiple recursion calls

def fibo(n):
    if(n<=1):
        return n
    last = fibo(n-1)
    slast = fibo(n-2)
    return last + slast

print(fibo(4))
    
```
