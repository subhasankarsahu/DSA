```python
n = int(input())

arr = list(map(int,input().split()))

# Create an array of size 13 filled with 0
hash_arr = [0] * 13

# Store frequencies
for num in arr:
    hash_arr[num] += 1

q = int(input())

for i in range(q):
    number = int(input())
    print(hash_arr[number])
```

```python
# Integer Hashing using dictinary(map)

n = int(input())

arr = list(map(int,input().split()))

mpp = {}

for num in arr:
    if num in mpp:
        mpp[num] += 1
    else:
        mpp[num] = 1

q = int(input())

for i in range(q):
    number = int(input())

    if number in mpp:
        print(mpp[number])
    else:
        print(0)
    
```

```python
# Character Hashing

s = input()

hash_arr = [0] * 26

for ch in s:
    index = ord(ch) - ord('a')
    hash_arr[index] += 1

q = int(input())

for i in range(q):
    c = input()

    index = ord(c) - ord('a')
    print(hash_arr[index])
```

```python
n = int(input())
arr = list(map(int,input().split()))

freq = {}

for num in arr:
    freq[num] = freq.get(num, 0) + 1

q = int(input())

for _ in range(q):
    number = int(input())
    print(freq.get(number, 0))
```
