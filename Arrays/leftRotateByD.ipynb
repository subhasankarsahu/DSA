```python
# In-Place Reversal Alg

def reverse(arr, start, end):
    while start < end:
        arr[start], arr[end] = arr[end], arr[start]

        start += 1
        end -= 1

def leftRotate(arr, d):
    n = len(arr)

    if n == 0:
        return arr

    d = d % n
    reverse(arr, 0, d-1)
    reverse(arr, d, n-1)
    reverse(arr, 0, n-1)
    return arr

arr = [1, 2, 3, 4, 5, 6, 7]
leftRotate(arr, 5)
print(arr)
```

```python
# Slicing

def leftRotate(arr, d):
    if not arr:
        return arr

    d = d % len(arr)
    return arr[d:] + arr[:d]

arr = [1, 2, 3, 4, 5, 6, 7]
print(leftRotate(arr, 5))
```
