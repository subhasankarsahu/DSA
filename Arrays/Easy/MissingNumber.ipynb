```python
# Brute Force - Linear Search

def missinNumber(nums: list[int]) -> int:
    n = len(nums)

    for i in range(n + 1):
        if i not in nums:
            return i

    return -1
```

```python
# Better : Sorting

def missinNumber(nums: list[int]) -> int:
    nums.sort()
    for i in range(len(nums)):
        if nums[i] != i:
            return i
    return len(nums)
```

```python
# Better 2 : Hashing

def missinNumber(nums: list[int]) -> int:
    num_set = set(nums)
    for i in range(len(nums) + 1):
        if i not in num_set:
            return i
    return -1
```

```python
# Optimal 1 : Sum

def missinNumber(nums: list[int]) -> int:
    n = len(nums)
    expected_sum = n * (n + 1) / 2
    return expected_sum - sum(nums)
```

```python
# Optimal 2 : XOR

def missinNumber(nums: list[int]) -> int:
    missing = len(nums)
    for i, num in enumerate(nums):
        missing ^= i ^ num
    return missing
```
