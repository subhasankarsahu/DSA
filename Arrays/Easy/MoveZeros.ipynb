```python
# In-Place with Two Pointers

def moveZeroes(arr):
    pos = 0

    for i in range(len(arr)):
        if arr[i] != 0:
            arr[pos], arr[i] = arr[i], arr[pos]
            pos += 1

    return arr

arr = [0, 1, 0, 3, 12]
print(moveZeroes(arr))
```

```python
# List comprehension

def moveZeroes(arr):
    non_zeroes = [x for x in arr if x != 0]
    zeros = [0] * (len(arr) - len(non_zeroes))

    return non_zeroes + zeros

arr = [0, 1, 0, 3, 12]
print(moveZeroes(arr))
```
