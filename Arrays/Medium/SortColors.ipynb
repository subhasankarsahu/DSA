```python
# Brute : Sorting

def sortColors(nums: list[int]) -> None:
    nums.sort()

    return nums

nums = [2,0,2,1,1,0]
print(sortColors(nums))
```

```python
# Better Approach (Counting / Two-Pass)

def sortColors(nums: list[int]) -> None:
    count_0 = count_1 = count_2 = 0
    
    for num in nums:
        if num == 0:
            count_0 += 1
        elif num == 1:
            count_1 += 1
        else:
            count_2 += 1
            
    for i in range(len(nums)):
        if i < count_0:
            nums[i] = 0
        elif i < count_0 + count_1:
            nums[i] = 1
        else:
            nums[i] = 2
    return nums

nums = [2,0,2,1,1,0]
print(sortColors(nums))
```

```python
# Optimal Approach (Dutch National Flag Algorithm)

def sortColors_optimal(nums: list[int]) -> None:
    low, mid, high = 0, 0, len(nums) - 1
    
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:  # nums[mid] == 2
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1

    return nums

nums = [2,0,2,1,1,0]
print(sortColors(nums))
```
