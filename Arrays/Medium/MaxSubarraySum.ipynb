```python
# Brute 

class Solution:
    def maxSubArray(self, nums: list[int]) -> int:
        max_sum = float('-inf')
        n = len(nums)
        for i in range(n):
            for j in range(i, n):
                current_sum = 0
                for k in range(i, j + 1):
                    current_sum += nums[k]
                max_sum = max(max_sum, current_sum)
        return max_sum
```

```python
# Optimal : Kadane's Algorithm

class Solution:
    def maxSubArray(self, nums: list[int]) -> int:
        max_sum = float('-inf')
        current_sum = 0
        for num in nums:
            current_sum += num
            max_sum = max(max_sum, current_sum)
            if current_sum < 0:
                current_sum = 0
        return max_sum
```
