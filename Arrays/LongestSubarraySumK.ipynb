```python
# Brute Force

def longest_subarray(nums: list[int], k: int) -> int:
    n = len(nums)
    maxLen = 0

    for i in range(n):
        current_sum = 0
        for j in range(i, n):
            current_sum += nums[j]
            if current_sum == k:
                maxLen = max(maxLen, j - i + 1)

    return maxLen

nums = [1,2,3,1,1,1,1,2,3,1]
k = 6
print(longest_subarray(nums, k))
```

```python
# Better (Prefix sum + Hash map)

def longest_subarray(nums: list[int], k: int) -> int:
    prefix_map = {}
    current_sum = 0
    maxLen = 0

    for i, num in enumerate(nums):
        current_sum += num

        if current_sum == k:
            maxLen = i + 1

        rem = current_sum - k
        if rem in prefix_map:
            maxLen = max(maxLen, i - prefix_map[rem])

        if current_sum not in prefix_map:
            prefix_map[current_sum] = i

    return maxLen

nums = [1,2,3,1,1,1,1,2,3,1]
k = 6
print(longest_subarray(nums, k))
```

```python
# Optimal (Two Pointers / Sliding Window) only for non-negative numbers


def longest_subarray(nums: list[int], k: int) -> int:
    left = 0
    current_sum = 0
    maxLen = 0
    n = len(nums)

    for right in range(n):
        current_sum += nums[right]

        while left <= right and current_sum > k:
            current_sum -= nums[left]
            left += 1

        if current_sum == k:
            maxLen = max(maxLen, right - left + 1)

    return maxLen

nums = [1,2,3,1,1,1,1,2,3,1]
k = 6
print(longest_subarray(nums, k))
```
