```python
# Brute Force

def longest_subarray(nums: list[int], k: int) -> int:
    n = len(nums)
    maxLen = 0

    for i in range(n):
        current_sum = 0
        for j in range(i, n):
            current_sum += nums[j]
            if current_sum == k:
                maxLen = max(maxLen, j - i + 1)

    return maxLen

nums = [1,2,3,1,1,1,1,2,3,1]
k = 6
print(longest_subarray(nums, k))
```
